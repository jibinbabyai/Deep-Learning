```python
import cv2
import matplotlib.pyplot as plt
```

```python
img = cv2.imread(r"C:\Users\jibin\Desktop\PythonProject\dl\open-cv\face detection\jesus 2.jpg")

img1 = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
plt.imshow(img1)
plt.show()
```

```python
con_img = cv2.cvtColor(img1, cv2.COLOR_BGR2RGB)
plt.imshow(con_img)
plt.show()
```

```python
face_data = cv2.CascadeClassifier("haarcascade_frontalface_default.xml")
face_data
```

```python
faces = face_data.detectMultiScale(con_img,minNeighbors=4)
faces
```

```python
for (x,y,w,h) in faces:
    cv2.rectangle(con_img,(x,y),(x+w,y+h),color=(255,0,3))
plt.imshow(con_img)    
```
