```python
import cv2
import matplotlib.pyplot as plt
```

```python
eye_cas = cv2.CascadeClassifier("haarcascade_eye.xml")
eye_cas
```

```python
video =cv2.VideoCapture(0)

while True:
    sucess, img = video.read()
    gray_img = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
    eyes = eye_cas.detectMultiScale(gray_img, minNeighbors=10)
    for (x,y,w,h) in eyes:
        cv2.rectangle(img,(x,y),(x+w,y+h),(0,0,255),2)
    cv2.imshow("Eye",img)
    if cv2.waitKey(1) & 0xFF == ord("q"):
        break
video.release()
cv2.destroyAllWindows()
```

```python
video =cv2.VideoCapture("r.mp4")

while True:
    sucess, img = video.read()
    if sucess:
        gray_img = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
        eyes = eye_cas.detectMultiScale(gray_img, minNeighbors=15)
        for (x,y,w,h) in eyes:
            cv2.rectangle(img,(x,y),(x+w,y+h),(0,0,255),2)
        cv2.imshow("Eye",img)
        if cv2.waitKey(1) & 0xFF == ord("q"):
            break
    else:    
        print("Finished")
        break
video.release()
cv2.destroyAllWindows()
```
