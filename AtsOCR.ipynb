```python
img = cv2.imread(r'OCR\image.png')
```

```python
import matplotlib.pyplot as plt 
```

```python
data = pytesseract.image_to_data(img,output_type=pytesseract.Output.DICT)
```

```python
print(data.keys())
```

```python
n = len(data['text'])
n
```

```python
keywords = ['opencv','nlp','ml','ai','python','dl','scikitlearn']
import re
stats = 0
```

```python
for i in range(n):
    if data['text'][i].lower() in keywords:
        stats += 1
        x,y,w,h = data['left'][i],data['top'][i],data['width'][i],data['height'][i]
        cv2.rectangle(img,(x,y),(x+w,y+h),(0,0,255),1)
```

```python
cv2.imshow('Image',img)
cv2.waitKey(5000)
cv2.destroyAllWindows()
```

```python
percentage = (stats/len(keywords))*100
percentage
```

```python
if percentage > 75:
    print('Candidate Approved!!')
else:
    print('Candidate Rejected!!')
```
