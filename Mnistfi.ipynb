```python
from tensorflow.keras.datasets import mnist
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Conv2D  #convolutional layer importing
from tensorflow.keras.layers import MaxPool2D # pooling layer importing
from tensorflow.keras.layers import Flatten
from tensorflow.keras.layers import Dense
```

```python
(x_train,y_train),(x_test,y_test)=mnist.load_data()
```

```python
x_train.shape
```

```python
x_test.shape
```

```python
y_train.shape
```

```python
y_test.shape
```

```python
#reshape
x_train=x_train.reshape(60000,28,28,1)
x_test=x_test.reshape(10000,28,28,1)
```

```python
x_train.shape
```

```python
x_test.shape
```

```python
#normalization of x_train and x_test
x_train=x_train/255
x_test=x_test/255
```

```python
#defining model
model=Sequential()
#adding convolutional layer
model.add(Conv2D(32,(3,3),activation='relu',input_shape=(28,28,1)))
#adding pooling layer
model.add(MaxPool2D(2,2))
#added full connected layer
model.add(Flatten())
model.add(Dense(100,activation='relu'))
#adding output layer
model.add(Dense(10,activation='softmax'))
```

```python
model.compile(optimizer='adam',loss='sparse_categorical_crossentropy',metrics=['accuracy'])
```

```python
model.fit(x_train,y_train,epochs=10,batch_size=1000)
```

```python
model.evaluate(x_test,y_test,batch_size=1000)
```
