```python
#Hand landmark detection

#install mediapipe
#pip install mediapipe
import mediapipe as mp
import cv2
```

```python
#Hand Landmark detection
#Two modules -solutions and utils .submodules 
mphands = mp.solutions.hands #Hand detection
mp_drwaing = mp.solutions.drawing_utils #Draw the line for the connection between landmarks
```

```python
#creating obeject for the class hands
data = mphands.Hands()
```

```python
#opening webcam
video = cv2.VideoCapture(0)
while True:
    suc,img = video.read()
    img1 = cv2.cvtColor(img,cv2.COLOR_BGR2RGB)
    result = data.process(img1)
    #print(result.multi_hand_landmarks) #to detect the landmark.All landmarks having x,y,z
    if result.multi_hand_landmarks: #draw only if hand detect landmarks
        for x in result.multi_hand_landmarks: # for 21 landmarks
            mp_drwaing.draw_landmarks(img,x,mphands.HAND_CONNECTIONS,mp_drwaing.DrawingSpec(color=(255,0,00),thickness=5,circle_radius=5),mp_drwaing.DrawingSpec(color=(0,0,255),thickness=5))



    cv2.imshow('hands',img)
    if cv2.waitKey(1) & 0XFF == ord('q'):
        break

video.release()
cv2.destroyAllWindows()
```
