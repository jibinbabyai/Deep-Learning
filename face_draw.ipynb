```python
import cv2
import matplotlib.pyplot as plt
```

```python
img = cv2.imread("dog.jpg")
con_img= cv2.cvtColor(img, cv2.COLOR_BGR2RGB)  
print(con_img)
```

```python
plt.imshow(con_img)
plt.show()
```

```python
# bounding box
# 1) rectangle (need left, top, width)
# 2) circle (need center, radius)

cv2.rectangle(con_img, pt1=(60,40), pt2=(100,70), color=(255,0,0), thickness=3)
plt.imshow(con_img)
plt.show()
```

```python
img2 = cv2.imread("baby.jpg")
con_img2 = cv2.cvtColor(img2, cv2.COLOR_BGR2RGB)
plt.imshow(con_img2)
plt.show()
```

```python
cv2.circle(con_img2,center=(90,40),radius=20,color=(0,0,255),thickness=2)
plt.imshow(con_img2)
```

```python
cv2.putText(con_img2,text = "CUTE BABY",org = (0,80),fontFace = cv2.FONT_HERSHEY_SIMPLEX,fontScale=1,color = (255,255,255),thickness = 1)
plt.imshow(con_img2)
plt.show()
```
