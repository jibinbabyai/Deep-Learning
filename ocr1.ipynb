```python
import cv2
import pytesseract
```

```python
img = cv2.imread('ocr1.jpeg')
img
```

```python
cv2.imshow("OCR",img)
cv2.waitKey(5000)
cv2.destroyAllWindows()
```

```python
pytesseract.pytesseract.tesseract_cmd = r"c:\Program Files\Tesseract-OCR\tesseract.exe"
```

```python
# set path to ocr
text = pytesseract.image_to_string(img)
print(text)
```
