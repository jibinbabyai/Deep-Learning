```python
import ultralytics
model = ultralytics.YOLO('yolov8n.pt')
```

```python
import cv2
from google.colab.patches import cv2_imshow
result=model(["/content/drive/MyDrive/Colab Notebooks/car.jpg","/content/drive/MyDrive/Colab Notebooks/train.jpg","/content/drive/MyDrive/Colab Notebooks/bus.jpg"])

for i in result:
  i.save(filename="result.jpg")
  img=cv2.imread("result.jpg")
  cv2_imshow(img)
```
